```python
import pandas as pd
import matplotlib.pyplot as plt


co2_data = {
    "Year": [2000, 2005, 2010, 2015, 2020, 2025],
    "CO2_Emissions": [25000, 27000, 30000, 32000, 31000, 29000]
}

co2_df = pd.DataFrame(co2_data)
co2_df.to_csv("co2_emissions.csv", index=False)

print("co2_emissions.csv created!")

co2 = pd.read_csv("co2_emissions.csv")

plt.figure(figsize=(8,5))
plt.plot(co2["Year"], co2["CO2_Emissions"], marker='o', linestyle='-', color='green')
plt.title("Global CO2 Emissions Over Time", fontsize=14)
plt.xlabel("Year")
plt.ylabel("CO2 Emissions (Million Metric Tons)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()


if co2["CO2_Emissions"].iloc[-1] > co2["CO2_Emissions"].iloc[0]:
    print("Analysis: CO2 emissions have increased over the years.")
else:
    print("Analysis: CO2 emissions have decreased after peaking in 2015.")



```

```python
scores_data = {
    "Exam": [1, 2, 3, 4, 5, 6],
    "Score": [55, 60, 65, 70, 72, 78]
}

scores_df = pd.DataFrame(scores_data)
scores_df.to_csv("student_scores.csv", index=False)

print("student_scores.csv created!")

scores = pd.read_csv("student_scores.csv")

plt.figure(figsize=(8,5))
plt.plot(scores["Exam"], scores["Score"], marker='o', linestyle='-', color='blue')
plt.title("Student Performance Over Exams", fontsize=14)
plt.xlabel("Exam Number")
plt.ylabel("Score")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()


if scores["Score"].iloc[-1] > scores["Score"].iloc[0]:
    print("Insight: The student’s performance improved steadily over time.")
else:
    print("Insight: The student’s performance declined or remained stagnant.")
```
