```python
import pandas as pd

# Step 1: Create a sample dataset
data = {
    "User_ID": [101, 102, 103, 104, 105],
    "Course_Name": [
        "Python for Data Science",
        "Web Development",
        "Python for Data Science",
        "Machine Learning",
        "Python for Data Science"
    ],
    "Rating": [5, None, 4, 3, 2],   # Notice: one missing rating
    "Feedback": [
        "Great course, very useful!",
        "Loved the projects.",
        "Good explanation, clear concepts.",
        "Too fast paced.",
        "Needs more exercises."
    ],
    "Date": ["2025-08-20", "2025-08-21", "2025-08-22", "2025-08-23", "2025-08-24"]
}

# Convert dictionary to DataFrame
df = pd.DataFrame(data)

print("Original Data:\n", df)

# Step 2: Remove rows with missing Rating
df = df.dropna(subset=["Rating"])

# Step 3: Filter feedback for "Python for Data Science" with Rating >= 4
filtered = df[(df["Course_Name"] == "Python for Data Science") & (df["Rating"] >= 4)]

# Step 4: Display usernames (User_ID) and feedback
print("\nMeaningful Feedbacks:\n")
print(filtered[["User_ID", "Feedback"]])
```

```python
import pandas as pd

# Step 1: Create a sample dataset
data = {
    "Location": ["Delhi", "Mumbai", "Chennai", "Kolkata", "Bengaluru"],
    "PM2.5": [180, 90, 160, 200, None],
    "PM10": [250, 120, 300, None, 180],
    "NO2": [85, 60, 95, 110, None],
    "SO2": [25, 20, 30, 40, 15],
    "Date": ["2025-08-20", "2025-08-21", "2025-08-22", "2025-08-23", "2025-08-24"]
}

# Convert dictionary to DataFrame
df = pd.DataFrame(data)

print("Original Data:\n", df)

# Step 2: Fill missing pollutant values using forward fill
df = df.fillna(method="ffill")

# Step 3: Filter records with PM2.5 > 150 and NO2 > 80
polluted_days = df[(df["PM2.5"] > 150) & (df["NO2"] > 80)]

print("\nPolluted Day Records (PM2.5 > 150 and NO2 > 80):\n")
print(polluted_days[["Location", "PM2.5", "PM10", "NO2", "SO2", "Date"]])
```
